# Document loader in langchain

Documents ko read karke LangChain ke format mein convert karna.

Jaise hum insaan PDF khol kar padhte hain, waise hi LangChain ko bhi PDF, Word, Text file, Website, CSV padhni hoti hai.

Ye kaam Document Loader karta hai.

# Example

Socho tumhare paas ek PDF hai:

company_policy.pdf

LLM directly PDF nahi padh sakta.

Pehle PDF ko open karna padega.

PDF
↓
Document Loader
↓
Text Extract
↓
LangChain Document

Yehi Document Loader ka kaam hai.

# Example

Maan lo file hai:

AI is changing the world.

Machine Learning is part of AI.

Document Loader isse convert karega:

Document(
    page_content="AI is changing the world.
     
    Machine Learning is part of AI."
)

Ab LangChain is document ko process kar sakta hai.

# Text Loader (.txt)

Text Loader ek LangChain component hai jo .txt file ko read karta hai aur uske content ko LangChain ke Document object mein convert karta hai.

Simple words mein:

notes.txt
     ↓
TextLoader
     ↓
Document Object

# Example

Maan lo ek file hai:

notes.txt

Artificial Intelligence is the simulation of human intelligence.

Machine Learning is a subset of AI.

Ye text file hai.

LangChain directly file ko process nahi karega.

Pehle Text Loader file ko read karega.

notes.txt
     ↓
TextLoader
     ↓
Text Extract
     ↓
Document


```python
%%writefile notes.txt
Artificial Intelligence is the simulation of human intelligence.

Machine Learning is a subset of AI.

!pip install langchain_community
from langchain_community.document_loaders import TextLoader

loader = TextLoader("notes.txt")

documents = loader.load()

print(documents)
```

# What is Document Object?

LangChain har document ko ek special format mein rakhta hai.

Document(
    page_content="AI is changing the world",

    metadata={"source":"notes.txt"}
)

# page_content

Actual text.

docs[0].page_content

Output:

AI is changing the world.

Machine Learning is part of AI.

# metadata

Extra information.

docs[0].metadata

Output:

{
  'source': 'notes.txt'
}

# Directory Loader


DirectoryLoader ek LangChain loader hai jo ek folder (directory) ke andar ki multiple files ko load karta hai.

# Simple definition:

Ek folder ke andar jitni files hain, unhe automatically read karke Document objects mein convert kar deta hai.

# Problem Kya Thi?

Maan lo tumhare paas 100 files hain:

data/
 ├── ai.txt
 ├── ml.txt
 ├── dl.txt
 ├── nlp.txt
 └── cv.txt

Without DirectoryLoader:

loader1 = TextLoader("ai.txt")
loader2 = TextLoader("ml.txt")
loader3 = TextLoader("dl.txt")
loader4 = TextLoader("nlp.txt")
loader5 = TextLoader("cv.txt")

100 files hongi to 100 baar loader banana padega 😅

loader = DirectoryLoader(
    "data",

    glob="*.txt"
)



# WebBase Loader

# Output Structure
[
 Document(

   metadata={
      "source":"https://example.com"
   }
 )
]



# CSV Loader

CSV = Comma Separated Values

Ye ek file format hai jisme data table ki form mein store hota hai.

employees.csv

Name,Age,Department

Rahul,25,IT

Amit,30,HR

Priya,28,Finance

Excel sheet jaisa data, bas text format mein.

CSVLoader ek LangChain Document Loader hai jo CSV file ko read karta hai aur har row ko ek Document object mein convert karta hai.

# Actual Output
[
 Document(
   page_content=
   "Name: Rahul\nAge: 25\nDepartment: IT",

   metadata={
      "source":"employees.csv",
      "row":0
   }
 ),

 Document(
   page_content=
   "Name: Amit\nAge: 30\nDepartment: HR",

   metadata={
      "source":"employees.csv",
      "row":1
   }
 )
]
