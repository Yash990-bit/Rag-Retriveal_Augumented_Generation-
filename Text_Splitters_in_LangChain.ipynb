# Text Splitters

Text Splitter ek LangChain component hai jo bade document ko chhote-chhote pieces (chunks) mein todta hai.

# Why Need Text Splitter?

Maan lo PDF mein 500 pages hain.

Isliye document ko tod dete hain.

# Chunk Size Kya Hota Hai?

Ek chunk mein kitne characters ya tokens honge.

chunk_size = 1000

Matlab:

Maximum 1000 characters
per chunk

# Chunk Overlap

Maan lo:

Chunk 1:
Python is used in AI and Machine Learning.

Chunk 2:
Machine Learning is a subset of AI.

Agar split galat jagah hua to context toot sakta hai.

Isliye overlap use karte hai

# Structure-Based Text Splitting

Length-based splitting mein:

* Sirf length dekhi jati hai

Lekin Structure-Based Splitting mein:

* Document ki structure dekhi jati hai

Jaise:

Heading,
Subheading,
Paragraph,
Section,
Markdown Structure,
HTML Tags

ke basis par split hota hai.

# Simple Example

Document:

# Introduction

AI is changing the world.

# Machine Learning

Machine Learning is a subset of AI.

# Deep Learning

Deep Learning uses neural networks.

# Length-Based Splitting

Agar 50 characters par split kiya:

Chunk 1:
# Introduction
AI is changing the world.

# Mac

Chunk 2:
hine Learning
Machine Learning is a subset

❌ Heading toot gayi.

❌ Meaning toot gaya.

# Structure-Based Splitting
Chunk 1:
# Introduction

AI is changing the world.

Chunk 2:
# Machine Learning

Machine Learning is a subset of AI.

Chunk 3:
# Deep Learning

Deep Learning uses neural networks.

✅ Proper sections preserve hue.

# Document Structure-Based

Is technique mein document ke natural structure ke basis par chunks banaye jate hain.

Yaani splitter document ke organization ko samajhta hai:

Book
 ├─ Chapter 1
 │   ├─ Section 1.1
 │   ├─ Section 1.2
 │
 ├─ Chapter 2
 │   ├─ Section 2.1
 │   ├─ Section 2.2

Aur usi structure ke hisab se split karta hai.

# Document:

Employee Handbook

Chapter 1: Company Introduction

Welcome to our company.

Chapter 2: Leave Policy

Employees get 20 paid leaves.

Chapter 3: Work From Home

Employees can work remotely.

# Document Structure-Based Splitting

Output:

Chunk 1:
Chapter 1: Company Introduction

Welcome to our company.

--------------------

Chunk 2:
Chapter 2: Leave Policy

Employees get 20 paid leaves.

--------------------

Chunk 3:
Chapter 3: Work From Home

Employees can work remotely.

Notice:

✅ Chapter intact hai

✅ Meaning intact hai

✅ Context preserve hua

# Why Is It Better?

Question:

Leave policy kya hai?

Retriever ko directly:

Chapter 2: Leave Policy

mil jayega.

Instead of searching random chunks.

# Real-Life Analogy

Book ko padhte waqt:

❌ Random 500 words nahi padhte

✅ Chapter-wise padhte ho

Document Structure Splitting bhi exactly yehi karta hai.

# Sabse Simple Definition

Document Structure-Based Splitting matlab document ko uske headings, chapters aur sections ke hisaab se todna, taaki meaning aur context bana rahe. 🚀

# Ek student ko main aise bolunga:

"Jaise book ko chapter-wise padhte ho, waise hi AI document ko chapter-wise todta hai. Isko Document Structure-Based Splitting kehte hain."

# Semantic Meaning Based Structure

Is technique mein splitter sirf length ya heading nahi dekhta.

Ye dekhta hai:

Kis sentence ka meaning similar hai?

Aur similar meaning wale sentences ko ek chunk mein rakhta hai.

# Comparison
Type	           ---      Kaise Split Karta Hai
_____________________________________________

Length-Based	     ---      Character/Token count
______________________________________________

Structure-Based	     ---      Heading/Chapter
______________________________________________

Semantic-Based	     ---      Meaning/Topic


# Example

Document:

Python is a programming language.

Python is easy to learn.

Python is used in AI.

The Taj Mahal is in Agra.

Millions of tourists visit the Taj Mahal every year.
# Length-Based Splitting
Chunk 1:
Python is a programming language.

Python is easy to learn.

Chunk 2:
Python is used in AI.

The Taj Mahal is in Agra.

Chunk 3:
Millions of tourists visit the Taj Mahal.

❌ Python aur Taj Mahal ek hi chunk mein aa gaye.

Meaning mix ho gaya.

# Semantic Splitting
Chunk 1:
Python is a programming language.

Python is easy to learn.

Python is used in AI.

Chunk 2:
The Taj Mahal is in Agra.

Millions of tourists visit the Taj Mahal every year.

✅ Similar meaning ek saath.

✅ Different topic alag.
