# Retriever Types

# 1. Similarity Retriever

Sabse common aur basic retriever.

Kaam

Question aur chunks ke embeddings compare karta hai.

Question
    ↓
Similarity Search
    ↓
Top Similar Chunks
Example

Question:

What is Python?

Chunk:

Python is a programming language.

Sabse similar chunk return karega.

# 2. MMR Retriever

MMR = Maximum Marginal Relevance

Problem

Similarity search kabhi-kabhi same type ke chunks la deta hai.

Example:

Chunk 1 → Python basics

Chunk 2 → Python basics

Chunk 3 → Python basics

Sab same information.

Solution

MMR bolta hai:

Similar bhi lao
+
Different bhi lao

Output:

Chunk 1 → Python basics

Chunk 2 → Python in AI

Chunk 3 → Python libraries

More diverse results.

# Why MMR ?
MMR isliye use karte hain taaki same type ki information baar-baar na aaye aur alag-alag useful information mile.


# 3. Multi Query Retriever
Kaam

Ek question ko multiple questions mein convert karta hai.

Example:

User:

What is RAG?

AI bana sakta hai:

Explain RAG

How does RAG work?

RAG architecture

RAG pipeline

Fir sab queries search hoti hain.

Result better aata hai.

# 4. Contextual Compression Retriever
Problem

Kabhi retriever bahut bada text la deta hai.

Example:

10 pages ka chunk

LLM ko pura dena expensive hai.

Solution

Compression Retriever:

Important part rakho

Unnecessary part hatao

Output:

Relevant 2 paragraphs only

# 5. Parent Document Retriever
Problem

Chunking ke baad context lose ho sakta hai.

Example:

Chunk:
"20 paid leaves"

Question:

Leave policy kya hai?

Sirf chunk se context kam ho sakta hai.

Solution

Retriever:

Small chunk search karega

Lekin

Bada parent document return karega

# 6. Self Query Retriever

Bahut powerful.

LLM query ko samajhta hai aur filter banata hai.

Example metadata:

{
  "year": 2025,
  "department": "HR"
}

User:

Show HR policies from 2025

Retriever khud filter create karega:

department = HR

year = 2025

Aur matching documents laega.

# Wikipedia Retriever

Wikipedia Retriever ek retriever hai jo Wikipedia se relevant information dhundhkar laata hai.

# How it work


Maan lo user poochta hai:

What is Artificial Intelligence?
# Step 1: User Question
What is Artificial Intelligence?

↓

Wikipedia Retriever ko milta hai.

# Step 2: Wikipedia Search

Retriever Wikipedia mein search karta hai:

Artificial Intelligence

Jaise hum Google ya Wikipedia par search karte hain.

↓

Wikipedia article mil jata hai.

# Step 3: Relevant Content Extract

Article se useful text nikala jata hai.

Example:

Artificial Intelligence (AI) is a field
of computer science that focuses on
creating systems capable of performing
tasks that normally require human intelligence.

↓

Document format mein convert.

# Step 4: Return Documents
Document(
   page_content="Artificial Intelligence is..."
)

↓

Retriever ye content LLM ko de deta hai.

# Vector Store Retriever

Vector Store Retriever ek helper hai jo Vector Store ke andar se user ke question ke liye sabse relevant information dhundhkar laata hai.

# Why Need It?

Agar 10,000 chunks stored hain:

Chunk 1
Chunk 2
...
Chunk 10000

LLM sab nahi padh sakta.

Retriever bolta hai:

Main sirf useful chunks launga

# Vector Store Retriever Kaise Kaam Karta Hai?

Socho tumhare paas ek notebook hai jisme 1000 pages hain.

Page 1 → Cricket

Page 2 → Python

Page 3 → AI

Page 4 → Finance

...

Ab tum poochte ho:

Python kya hai?

Kya AI 1000 pages ek-ek karke padhega?

❌ Nahi

Yahin Retriever ka kaam shuru hota hai.

# Step 1: Documents Store Hote Hain

Document:

Python is a programming language.

Embedding banti hai:

[0.12, 0.45, 0.89]

Aur Vector Store mein save ho jati hai.

Document
   ↓
Embedding
   ↓
Vector Store
# Step 2: User Question Aata Hai

User:

What is Python?
# Step 3: Question Ki Embedding Banti Hai

Question bhi numbers mein convert hota hai.

What is Python?

↓

[0.11, 0.44, 0.90]
# Step 4: Retriever Search Karta Hai

Retriever compare karta hai:

Question Vector

vs

Stored Vectors

Jaise:

Python Vector  ← Bahut Similar ✅

Cricket Vector ← Similar Nahi ❌

Finance Vector ← Similar Nahi ❌
# Step 5: Relevant Chunks Mil Jaate Hain

Retriever bolta hai:

Ye Python wala chunk sabse relevant hai.

Aur us chunk ko nikal leta hai.

Chunk:
Python is a programming language.
